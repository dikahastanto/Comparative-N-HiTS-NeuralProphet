```python
# ======================================
# 1. SETUP
# ======================================
import warnings
warnings.filterwarnings("ignore")

import pandas as pd

# ======================================
# 2. LOAD DATASET
# ======================================
file_path = "climate_data.csv"  # ganti sesuai file kamu
df = pd.read_csv(file_path)

# ======================================
# 3. FILTER STATION (JAKARTA - 96745)
# ======================================
df = df[df['station_id'] == 96745].copy()

# =========================
# CLEANING
# =========================
df['date'] = pd.to_datetime(df['date'], dayfirst=True)
df = df.sort_values('date')

# handle missing value
df = df.fillna(method='ffill').fillna(method='bfill')

# ======================================
# 5. AMBIL KOLOM (ds, y)
# ======================================
df_model = df[['date', 'Tavg']].copy()
df_model.columns = ['ds', 'y']

# ======================================
# 6. SAVE KE CSV
# ======================================
output_path = "climate_jakarta_96745.csv"
df_model.to_csv(output_path, index=False)

print("Selesai. File disimpan di:", output_path)
print(df_model.head())

# ambil dari kolom Tavg
```
