```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

# load data
df = pd.read_csv("climate_jakarta_96745.csv")

# parsing & sorting
df['ds'] = pd.to_datetime(df['ds'])
df = df.sort_values('ds')

# ====== BASIC STATS ======
length = len(df)
mean = df['y'].mean()
std = df['y'].std()
min_val = df['y'].min()
max_val = df['y'].max()
cv = std / mean
missing_pct = df['y'].isna().sum() / length * 100

# ====== FREQUENCY ======
freq = pd.infer_freq(df['ds'])
if freq is None:
    freq = "Daily"

# ====== OUTPUT ======
print("Frequency:", freq)
print("Length:", length)
print("Mean:", mean)
print("Std:", std)
print("Min:", min_val)
print("Max:", max_val)
print("CV:", cv)
print("Missing (%):", missing_pct)
```
