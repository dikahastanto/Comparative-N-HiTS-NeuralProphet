```python
import pandas as pd

# Load data
df = pd.read_csv("AAPL.csv")

# Rename kolom penting
df = df.rename(columns={
    "Date": "ds",
    "Close/Last": "y"
})

# Bersihin tanda $
df["y"] = df["y"].replace(r"[\$,]", "", regex=True)

# Convert tipe data
df["y"] = pd.to_numeric(df["y"], errors="coerce")

# Convert tanggal
df["ds"] = pd.to_datetime(df["ds"], format="%m/%d/%Y")

# Sorting + cleaning
df = df.sort_values("ds")
df = df.drop_duplicates()

# Missing value
df = df.fillna(method="ffill").fillna(method="bfill")

# Ambil hanya kolom yang dipakai
df = df[["ds", "y"]]

# Export ke CSV (FINAL DATASET)
df.to_csv("AAPL_clean.csv", index=False)

print(df.head())

# ambil dari kolom Close/Last
```
