```python
import numpy as np
import matplotlib.pyplot as plt

# load
actual = np.load("actual.npy")          # dari NeuralProphet
pred_np = np.load("pred_np.npy")
pred_nhits = np.load("pred_nhits.npy")

# samakan panjang
min_len = min(len(actual), len(pred_np), len(pred_nhits))

actual = actual[-min_len:]
pred_np = pred_np[-min_len:]
pred_nhits = pred_nhits[-min_len:]

# plot
plt.figure(figsize=(12,5))

plt.plot(actual, label="Actual")
plt.plot(pred_np, label="NeuralProphet")
plt.plot(pred_nhits, label="N-HiTS")

plt.legend()
plt.title("Prediction vs Actual")
plt.xlabel("Index")
plt.ylabel("Value")

plt.tight_layout()
plt.show()
```
