```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

# load data
df = pd.read_csv("oil_price.csv")

# parsing & sorting
df['observation_date'] = pd.to_datetime(df['observation_date'])
df = df.sort_values('observation_date')

# kolom target
target_col = 'DCOILWTICO'

# ====== BASIC STATS ======
length = len(df)
mean = df[target_col].mean()
std = df[target_col].std()
min_val = df[target_col].min()
max_val = df[target_col].max()
cv = std / mean
missing_pct = df[target_col].isna().sum() / length * 100

# ====== FREQUENCY ======
freq = pd.infer_freq(df['observation_date'])
if freq is None:
    freq = "Daily"

# ====== OUTPUT ======
print("Frequency:", freq)
print("Length:", length)
print("Mean:", mean)
print("Std:", std)
print("Min:", min_val)
print("Max:", max_val)
print("CV:", cv)
print("Missing (%):", missing_pct)
```
