```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

# load data
df = pd.read_csv("oil_price.csv")

# parsing & sorting
df['observation_date'] = pd.to_datetime(df['observation_date'])
df = df.sort_values('observation_date')

# kolom target
target_col = 'DCOILWTICO'

# ====== DISTRIBUTION STATS ======
skewness = df[target_col].skew()
kurtosis = df[target_col].kurtosis()

# ====== SIMPLE CHARACTER LABEL (optional) ======
if abs(skewness) > 1:
    skew_label = "Highly skewed"
elif abs(skewness) > 0.5:
    skew_label = "Moderately skewed"
else:
    skew_label = "Approximately symmetric"

if kurtosis > 3:
    kurt_label = "Heavy-tailed"
elif kurtosis > 0:
    kurt_label = "Moderate tail"
else:
    kurt_label = "Light-tailed"

# ====== OUTPUT ======
print("Skewness:", skewness)
print("Kurtosis:", kurtosis)
print("Skewness Label:", skew_label)
print("Kurtosis Label:", kurt_label)
```
