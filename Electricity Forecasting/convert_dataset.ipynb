```python
import pandas as pd

# 1. Load dataset
df = pd.read_csv("LD2011_2014.txt", sep=';', index_col=0)

# 2. Convert datetime
df.index = pd.to_datetime(df.index)

# 3. Fix format angka (kalau pakai koma)
df = df.replace(',', '.', regex=True).astype(float)

# 4. SUM semua MT
df['total_load'] = df.sum(axis=1)

# 5. Ambil total saja
df = df[['total_load']]

# 6. Resample ke daily
df_daily = df.resample('D').mean()

# 7. Format ke ds, y
df_daily = df_daily.reset_index()
df_daily.columns = ['ds', 'y']

# 8. Handle missing
df_daily = df_daily.fillna(method='ffill')

# 9. Simpan ke CSV
df_daily.to_csv("electricity_daily_total.csv", index=False)

print("✅ File berhasil disimpan: electricity_daily_total.csv")
```
