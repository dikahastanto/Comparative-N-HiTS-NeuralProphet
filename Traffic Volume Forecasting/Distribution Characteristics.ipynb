```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")

# load data
df = pd.read_csv("metro_interstate_traffic_volume_daily.csv")

# parsing & sorting
df['ds'] = pd.to_datetime(df['ds'])
df = df.sort_values('ds')

# ====== DISTRIBUTION ======
skewness = df['y'].skew()
kurtosis = df['y'].kurtosis()

# ====== OUTPUT ======
print("Skewness:", skewness)
print("Kurtosis:", kurtosis)
```
