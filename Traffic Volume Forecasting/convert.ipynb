```python
import pandas as pd

# Load dataset
df = pd.read_csv("metro_interstate_traffic_volume.csv")

# Pastikan kolom ds sebagai datetime
df["ds"] = pd.to_datetime(df["ds"])

# Jumlahkan traffic per hari
daily_df = df.groupby(df["ds"].dt.date)["y"].sum().reset_index()

# Ubah kolom ds jadi datetime (bukan date object) supaya NeuralProphet bisa pakai
daily_df["ds"] = pd.to_datetime(daily_df["ds"])

# Simpan ke CSV baru
daily_df.to_csv("metro_interstate_traffic_volume_daily.csv", index=False)

print(daily_df.head())

# yang dipake kolom traffic_volume di group by hari
```
